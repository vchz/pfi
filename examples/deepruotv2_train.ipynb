```python
import numpy as np
import torch
import sys
from pathlib import Path

deepruot_path = Path("/mnt/home/vchardes/DeepRUOTv2")
if str(deepruot_path) not in sys.path:
    sys.path.insert(0, str(deepruot_path))
    
from pfi.utils.data import load_data, X_from_snapshots
from pfi.flow.external import DeepRUOTv2

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
device


from examples.mpl_config import apply_mpl_config
apply_mpl_config()

```

```python
path = '/mnt/home/vchardes/ceph/datasets/HSC_data/10XChromiumV3_10.1038_s41467-021-27159-x_10.5281_zenodo.5291737_exvivo.h5ad'

seed = 0
np.random.seed(seed)
torch.manual_seed(seed)
nsamples = 6000
genes = ['gata1','fli1','klf1','gata2','gfi1','gfi1b','runx1','tal1','jun','spi1','zfpm1','lmo2','etv6','erg','cebpa',\
         'meis1','sall4','myc','foxo3','zbtb7a','nanog','nfe2','stat3','mef2c']

samples, times, ind_array, cell_types = load_data(
    path=path,
    nsamples=nsamples,
    genes=genes,
    time_key='day',
    cell_type_key='cell_type',
    seed=seed,
)

X = X_from_snapshots(samples, times)

print('samples shape:', samples.shape)
print('X shape:', X.shape)
print('times:', times)
```

```python
model = DeepRUOTv2(
    device=device,
    solver_kwargs=dict(
        exp_name='deepruotv2_smoke',
    ),
)

model.fit(X)
```
